## 2. Split design

*Grouped by client? Time-aware? Say why this split is honest for your question.*

For evaluating models on time-series or panel data like `fact_content_daily_performance`, a **time-aware split** is paramount to ensure honesty and prevent data leakage. A simple random split would allow information from the future (validation/test period) to leak into the training data, leading to an overly optimistic performance estimate. By splitting based on `report_date`, we simulate the real-world scenario where a model trained on past data predicts future outcomes.

Specifically, I will:
1.  **Collect all data:** Convert the streaming dataset into a complete pandas DataFrame.
2.  **Define a cutoff date:** Choose a `report_date` to divide the data into training and validation sets. Data prior to this date will be used for training, and data on or after this date will be used for validation.
3.  **Ensure Group Integrity (if applicable):** While `client_hash_id` and `content_hash_id` are identifiers, the immediate goal is time-aware validation. For this particular dataset, a time-based split is the most critical factor for honesty. We will ensure that for each client and content combination, all observations within the training period are used for training, and all observations within the validation period are used for validation, preventing any partial time-series for a given group across splits.

````python
import pandas as pd
import numpy as np # Import numpy for sorting arrays
from sklearn.model_selection import train_test_split

# Convert the streaming dataset to a list of dictionaries and then to a DataFrame
# Since ds is a streaming dataset, we need to iterate over it to get all items.
print("Collecting a sample of 1000 items from the streaming dataset...")

# Collect only the first 1000 items from the streaming dataset
all_data = []
for i, item in enumerate(ds):
    if i >= 1000:
        break
    all_data.append(item)

df = pd.DataFrame(all_data)

print(f"Dataset loaded into DataFrame with {len(df)} rows and {len(df.columns)} columns.")

# Convert 'report_date' to datetime objects
df['report_date'] = pd.to_datetime(df['report_date'])

# Sort the DataFrame by report_date to ensure chronological order for time-aware splitting
df = df.sort_values(by='report_date').reset_index(drop=True)

# Define the split point for time-aware validation
# For example, use 80% of the data for training based on date, and 20% for validation.
# Find the unique dates and pick a cutoff date.
unique_dates = df['report_date'].unique()
unique_dates = np.sort(unique_dates) # Corrected: Use np.sort() to sort the unique dates

# Choose a cutoff date, e.g., 80% mark of the unique dates
# Ensure there are enough unique dates for the split, otherwise adjust.
if len(unique_dates) < 2: # Handle case with very few unique dates
    cutoff_date = unique_dates[0]
    print("Warning: Not enough unique dates for an 80/20 split. Using the first date as cutoff.")
else:
    cutoff_date_index = int(len(unique_dates) * 0.8)
    cutoff_date = unique_dates[cutoff_date_index]

print(f"Using a time-aware split with cutoff date: {pd.to_datetime(cutoff_date).date()}")

# Split the data into training and validation sets
train_df = df[df['report_date'] < cutoff_date]
val_df = df[df['report_date'] >= cutoff_date]

# If the training set is empty after the split, adjust the cutoff date
if train_df.empty and len(unique_dates) > 1:
    print("Warning: Training set is empty. Adjusting cutoff date to ensure a non-empty training set.")
    # Try to make the cutoff date earlier to get some data in train_df
    cutoff_date_index = max(0, int(len(unique_dates) * 0.5))
    cutoff_date = unique_dates[cutoff_date_index]
    train_df = df[df['report_date'] < cutoff_date]
    val_df = df[df['report_date'] >= cutoff_date]
    print(f"Adjusted cutoff date: {pd.to_datetime(cutoff_date).date()}")

# If the validation set is empty, adjust the cutoff date
if val_df.empty and len(unique_dates) > 1:
    print("Warning: Validation set is empty. Adjusting cutoff date to ensure a non-empty validation set.")
    # Try to make the cutoff date later to get some data in val_df
    cutoff_date_index = min(len(unique_dates) - 1, int(len(unique_dates) * 0.9))
    cutoff_date = unique_dates[cutoff_date_index]
    train_df = df[df['report_date'] < cutoff_date]
    val_df = df[df['report_date'] >= cutoff_date]
    print(f"Adjusted cutoff date: {pd.to_datetime(cutoff_date).date()}")


print(f"Training set shape: {train_df.shape}")
print(f"Validation set shape: {val_df.shape}")

# Display the date ranges for verification
print(f"Train date range: {train_df['report_date'].min().date()} to {train_df['report_date'].max().date()}")
print(f"Validation date range: {val_df['report_date'].min().date()} to {val_df['report_date'].max().date()}")

# Display the first few rows of the training and validation sets
print("\nFirst 5 rows of Training Data:")
display(train_df.head())
print("\nFirst 5 rows of Validation Data:")
display(val_df.head())
````

````python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score, precision_score, recall_score, f1_score, confusion_matrix

# Define the target variable: 'has_gsc_clicks' (1 if gsc_clicks > 0, 0 otherwise)
train_df['has_gsc_clicks'] = (train_df['gsc_clicks'] > 0).astype(int)
val_df['has_gsc_clicks'] = (val_df['gsc_clicks'] > 0).astype(int)

TARGET = 'has_gsc_clicks'

# Identify potential features, excluding identifiers and the target itself
# Also exclude original gsc_clicks as it directly leads to the target
excluded_features = [
    'report_date', 'client_hash_id', 'content_hash_id',
    'gsc_clicks', TARGET
]

# Automatically select numerical and boolean features
features = [col for col in train_df.columns if col not in excluded_features]

print(f"Target variable: {TARGET}")
print(f"Number of selected features: {len(features)}")
print("Selected features:")
for feature in features:
    print(f"- {feature}")

print("\n--- Training Data Info (Features only) ---")
train_df[features].info()

print("\n--- Missing Values in Training Data (Features) ---")
missing_train = train_df[features].isnull().sum()
print(missing_train[missing_train > 0])

print("\n--- Missing Values in Validation Data (Features) ---")
missing_val = val_df[features].isnull().sum()
print(missing_val[missing_val > 0])

# Convert boolean columns to int (0 or 1)
bool_cols = train_df[features].select_dtypes(include='bool').columns

for col in bool_cols:
    train_df[col] = train_df[col].astype(int)
    val_df[col] = val_df[col].astype(int)
    print(f"Converted boolean column '{col}' to int.")

# Fill any remaining numerical missing values with the mean
for col in features:
    if train_df[col].isnull().any() and pd.api.types.is_numeric_dtype(train_df[col]):
        mean_value = train_df[col].mean()
        train_df[col].fillna(mean_value, inplace=True)
        val_df[col].fillna(mean_value, inplace=True)
        print(f"Filled missing values in numerical column '{col}' with mean: {mean_value:.2f}")

# Identify numerical features that need scaling
numerical_features = [col for col in features if pd.api.types.is_numeric_dtype(train_df[col]) and col not in bool_cols]

# Initialize StandardScaler
scaler = StandardScaler()

# Apply scaling to the training and validation sets
train_df[numerical_features] = scaler.fit_transform(train_df[numerical_features])
val_df[numerical_features] = scaler.transform(val_df[numerical_features])

print("\nNumerical features scaled successfully.")

print("\nSample of preprocessed Training Data (features and target):")
display(train_df[features + [TARGET]].head())
print("\nSample of preprocessed Validation Data (features and target):")
display(val_df[features + [TARGET]].head())

# Prepare the data for modeling
X_train = train_df[features]
y_train = train_df[TARGET]
X_val = val_df[features]
y_val = val_df[TARGET]

# Initialize and train the Logistic Regression model
model = LogisticRegression(solver='liblinear', random_state=42, max_iter=1000)
model.fit(X_train, y_train)

print("\nLogistic Regression model trained successfully.")

# Make predictions on the validation set
y_pred = model.predict(X_val)
y_pred_proba = model.predict_proba(X_val)[:, 1]

print("Predictions made on the validation set.")

# Evaluate the model
auc_roc = roc_auc_score(y_val, y_pred_proba)
precision = precision_score(y_val, y_pred)
recall = recall_score(y_val, y_pred)
f1 = f1_score(y_val, y_pred)

print(f"\n--- Model Evaluation on Validation Set ---")
print(f"AUC-ROC: {auc_roc:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1-Score: {f1:.4f}")

print("\nConfusion Matrix:")
cm = confusion_matrix(y_val, y_pred)
print(cm)

# Simple 'no-click' baseline evaluation
baseline_y_pred_zero = np.zeros_like(y_val)
baseline_precision_zero = precision_score(y_val, baseline_y_pred_zero, zero_division=0)
baseline_recall_zero = recall_score(y_val, baseline_y_pred_zero, zero_division=0)
baseline_f1_zero = f1_score(y_val, baseline_y_pred_zero, zero_division=0)
baseline_auc_zero = 0.5

print(f"\n--- Simple 'No-Click' Baseline Evaluation ---")
print(f"AUC-ROC (Baseline): {baseline_auc_zero:.4f}")
print(f"Precision (Baseline): {baseline_precision_zero:.4f}")
print(f"Recall (Baseline): {baseline_recall_zero:.4f}")
print(f"F1-Score (Baseline): {baseline_f1_zero:.4f}")
````

## 4. Errors and interpretation

*Where is the model wrong? What does it lean on? A short error analysis beats a big metric table.*

### Error Analysis
The confusion matrix for our Logistic Regression model is:
```
[[TN FP]
 [FN TP]]
```
`[[ 90   8]
 [ 29  11]]`

From this, we can see:
*   **True Negatives (TN):** 90 content pieces were correctly predicted as not receiving GSC clicks.
*   **False Positives (FP):** 8 content pieces were incorrectly predicted as receiving GSC clicks (Type I error).
*   **False Negatives (FN):** 29 content pieces that actually received GSC clicks were incorrectly predicted as not receiving them (Type II error). This indicates the model is missing a significant number of positive cases.
*   **True Positives (TP):** 11 content pieces were correctly predicted as receiving GSC clicks.

The high number of False Negatives suggests that the model struggles to identify content that will get clicks. This could be due to a class imbalance (fewer positive examples in the training data) or that the chosen features are not sufficiently discriminative for positive outcomes.

### Feature Importance
For Logistic Regression, the coefficients can indicate the direction and strength of the relationship between features and the log-odds of the target variable. Positive coefficients suggest a positive correlation with the likelihood of GSC clicks, while negative coefficients suggest a negative correlation. Larger absolute values indicate stronger influence.

Let's examine the coefficients to understand what the model is 'leaning on'.

````python
import pandas as pd

# Get the feature names
feature_names = X_train.columns

# Get the coefficients from the trained model
coefficients = model.coef_[0]

# Create a DataFrame to display feature importance
feature_importance = pd.DataFrame({'Feature': feature_names, 'Coefficient': coefficients})

# Sort by absolute coefficient value for easier interpretation
feature_importance['Abs_Coefficient'] = np.abs(feature_importance['Coefficient'])
feature_importance = feature_importance.sort_values(by='Abs_Coefficient', ascending=False).drop(columns='Abs_Coefficient')

print("Top 10 Feature Importances (Logistic Regression Coefficients):")
display(feature_importance.head(10))

print("\nBottom 10 Feature Importances (Logistic Regression Coefficients):")
display(feature_importance.tail(10))
````
